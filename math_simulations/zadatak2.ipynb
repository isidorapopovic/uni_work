Visok pritisak više sunca\
Nizak pritisak  ili sunce ili kiša\
Za rešavanje ovog problema koristićemo Markovljev lanac, gde temperatura sutra zavisi od svih temperatura do sada

```python
import numpy as np
import matplotlib.pyplot as plt
import random
import pandas as pd 
import os 
```

```python
P = np.array([
    [0.9,  0.0,  0.1],
    [0.05, 0.5,  0.45],
    [0.05, 0.7,  0.25]
])
```

```python
pi0 = np.array([0.10, 0.70, 0.20])
```

```python
states = ["V-sunčano", "N-sunčano", "N-kišovito"]
df_P = pd.DataFrame(P, index=states, columns=states)
df_pi0 = pd.DataFrame(pi0.reshape(1,-1), columns=states, index=["1. jun"])
```

```python
print("Tranziciona matrica P:")
display(df_P)
```

```python
print("Početna raspodela pi0 na 1. jun:")
display(df_pi0)
```

```python
def verovatnoca(pi_initial, P, n):
    Pn = np.linalg.matrix_power(P, n)  # P^n
    return pi_initial.dot(Pn)
```

```python
p6jun=verovatnoca(pi0, P, 5)
p1jul=verovatnoca(pi0, P, 30)
```

```python
print(f"Raspodela na 6. jun ({p6jun} dana kasnije):")
print(f"  V-sunčano:   {p6jun[0]:.2f}")
print(f"  N-sunčano:   {p6jun[1]:.2f}")
print(f"  N-kišovito:  {p6jun[2]:.2f}")

print(f"\nRaspodela na 1. jul ({p1jul} dana kasnije):")
print(f"  V-sunčano:   {p1jul[0]:.3f}")
print(f"  N-sunčano:   {p1jul[1]:.3f}")
print(f"  N-kišovito:  {p1jul[2]:.3f}")
```

```python
pi0 = np.array([0.1, 0.7, 0.2], dtype=float)

def raspodela_za(pi0, P, n):
    return pi0 @ np.linalg.matrix_power(P, n)

pi_5  = raspodela_za(pi0, P, 5)
pi_30 = raspodela_za(pi0, P, 30)

print("Za 5 dana :", np.round(pi_5, 4))
print("Za 30 dana:", np.round(pi_30, 4))
```

```python
for dan in range(1, 7):
    n = dan - 1
    pin = raspodela_za(pi0, P, n)
    df = pd.DataFrame([np.round(pin, 4)], index=[f"{dan}. jun"], columns=states)
    print(f"\n=== {dan}. jun — raspodela stanja ===")
    print(df.to_string())
```

Kako vreme prolazi, vidimo da se sve više približavamo uniformnoj raspodeli — verovatnoća za svako vreme iznosi oko 33%, jer nam ulazni podaci slabe i ne donose nikakvu novu informaciju. Ako je danas bilo sunčano, očekujemo da će i sutra biti sunčano. Promena verovatnoće tranzicije prati i verovatnoću pojave tog vremena. Na primer, ako smanjimo verovatnoću da visok pritisak rezultuje sunčanim vremenom, ne možemo biti sigurni da će sutra biti sunčano
