```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd 
import os 
```

Logistički niz: p_{n+1} = k * p_n * (1 - p_n)

p0 je velicina pocente populacije

k stopa rasta po generaciji


```python
def logisticki_niz(k, p0, n):
    if not (0 < p0 < 1):
        raise ValueError("p0 mora biti u intervalu (0, 1).")
    if not (0 < k <= 5):
        raise ValueError("k je uobičajeno u intervalu (0, 4].")

    p = [p0]
    for _ in range(n):
        p.append(k * p[-1] * (1 - p[-1]))
    return p  
```

```python
k = 4 
p0 = 0.10 # 0 < p0 < 1
n = 400   # broj koraka
```

```python
niz = logisticki_niz(k, p0, n)
```

```python
plt.figure()
plt.subplot(2,2,1)
plt.plot(range(len(niz)),logisticki_niz(1,0.10,400))
plt.xlabel("n")
plt.ylabel("p_n")
plt.title(f"Logistički niz: k=1, p0={p0}")
plt.grid(True)
plt.subplot(2,2,2)
plt.plot(range(len(niz)),logisticki_niz(2,0.10,400))
plt.xlabel("n")
plt.ylabel("p_n")
plt.title(f"Logistički niz: k=2, p0={p0}")
plt.grid(True)
plt.subplot(2,2,3)
plt.plot(range(len(niz)),logisticki_niz(3,0.10,400))
plt.xlabel("n")
plt.ylabel("p_n")
plt.title(f"Logistički niz: k=3, p0={p0}")
plt.grid(True)
plt.subplot(2,2,4)
plt.plot(range(len(niz)),logisticki_niz(4,0.10,400))
plt.xlabel("n")
plt.ylabel("p_n")
plt.title(f"Logistički niz: k=4, p0={p0}")
plt.grid(True)
plt.tight_layout()
```

Vidimo kako k, tj. stopa rasta, utiče na veličinu populacije — da se kreće od smanjenja, do stabilne konvergencije, do potpunih oscilacija

```python
n=30
k=2
p0=0.5
niz = logisticki_niz(k, p0, n)
plt.figure()
plt.subplot(2,1,1)
plt.plot(range(len(niz)),logisticki_niz(2,0.50,30))
plt.xlabel("n")
plt.ylabel("p_n")
plt.title("Logistički niz: k=2, p0=0.5")
plt.grid(True)
plt.subplot(2,1,2)
plt.plot(range(len(niz)),logisticki_niz(2.7,0.50,30))
plt.xlabel("n")
plt.ylabel("p_n")
plt.title("Logistički niz: k=2.7, p0=0.5")
plt.grid(True)
plt.tight_layout()
```
