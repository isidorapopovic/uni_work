T = 72/p\
The Rule of 72 is a formula used to calculate how long your investment will need to double in value, given its annual rate of return.

```python
import numpy as np
import matplotlib.pyplot as plt
```

Koristeći eksponencijalnu vezu izračunati tačno vreme t potrebno
da se vrednost deponovanog novca duplira.

```python
p = np.linspace(0.1, 20, 200)  
r = 1 + p/100   
```

```python
#V2 = V1 * r^t  =>  2 = r^t  =>  t_exact = ln(2) / ln(r)
t_exact = np.log(2) / np.log(r)
t_approx = 72 / p
```

Uporediti grafike za tačnu i približnu formulu.

```python
plt.figure(figsize=(8,5))
plt.plot(p, t_exact, label='Precizno: $t=\\ln2/\\ln(1+p/100)$')
plt.plot(p, t_approx, '--', label='Približno: $t=72/p$')
plt.xlabel('Kamatna stopa p (%)')
plt.ylabel('Vreme do udvostručenja t (god)')
plt.title('Uporedni prikaz tačne i približne formule')
plt.legend()
plt.grid()
```

Kako se menja veličina greške sa promenom kamate r

```python
greska = t_approx - t_exact

plt.figure()
plt.plot(p, greska)
plt.xlabel('Kamatna stopa p (%)')
plt.ylabel('Apsolutna greška $t_{approx}-t_{exact}$ (god)')
plt.title('Kako greška zavisi od kamatne stope')
plt.grid()
```

```python
sum(greska)
```

Sa grafika možemo zaključiti da se sa porastom vrednosti kamatne stope apsolutna greška smanjuje. Ono što je zanimljivo jeste da je za vrednosti kamatne stope veće od 5 greška prilično zanemarljiva, a otprilike toliko iznosi i prosečna kamatna stopa

Trenutni model je 72/p, ali mi zapravo hoćemo da izračunamo to 72, koje ćemo označiti sa C\
Zapravo, hoćemo da izračunamo C/p = ln(2)/ln(1 + p/100)\
Ono što možemo sledeće da uradimo jeste da razvijemo u red za malo p\

t = ln(2)/p/100 → t = 100·ln(2)/p → 69.3

Ovo daje odlične aproksimacije za veoma male kamate, ali nije dobro za veće kamate. Ovo se rešava time što ćemo koristiti viši red razvoja kako bismo dobili manje greške za veće kamate, veće od 5

```python
t69=69.3/p
```

```python
plt.plot(p, t_exact, label='Precizno: $t=\\ln2/\\ln(1+p/100)$')
plt.plot(p, t_approx, '--', label='Približno: $t=72/p$')
plt.plot(p, t69, '-', label='Približno: $t=69.3/p$')
plt.xlabel('Kamatna stopa p (%)')
plt.ylabel('Vreme do udvostručenja t (god)')
plt.title('Uporedni prikaz tačne i približne formule')
plt.legend()
plt.grid()
```
